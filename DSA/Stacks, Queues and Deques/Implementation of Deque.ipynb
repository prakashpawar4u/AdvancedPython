# Implementation of Deque

In this lecture we will implement our own Deque class!

## Methods and Attributes

* Deque() creates a new deque that is empty. It needs no parameters and returns an empty deque.
* addFront(item) adds a new item to the front of the deque. It needs the item and returns nothing.
* addRear(item) adds a new item to the rear of the deque. It needs the item and returns nothing.
* removeFront() removes the front item from the deque. It needs no parameters and returns the item. The deque is modified.
* removeRear() removes the rear item from the deque. It needs no parameters and returns the item. The deque is modified.
* isEmpty() tests to see whether the deque is empty. It needs no parameters and returns a boolean value.
* size() returns the number of items in the deque. It needs no parameters and returns an integer.

## Deque Implementation

```python
class Deque:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def addFront(self, item):
        self.items.append(item)

    def addRear(self, item):
        self.items.insert(0,item)

    def removeFront(self):
        return self.items.pop()

    def removeRear(self):
        return self.items.pop(0)

    def size(self):
        return len(self.items)
```

# 🛠 Deque Using Doubly Linked List (from scratch)
Step-by-Step Implementation

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.prev = None
        self.next = None

    
class Deque:
    def __init__(self):
        self.front = None
        self.rear = None
        self.count = 0

    
    def is_empty(self):
        return self.count == 0
    
    def size(self):
        return self.count
    
    def append_front(self, value):
        new_node = Node(value)
        if self.is_empty():
            self.front = self.rear = new_node
        
        else:
            new_node.next = self.front
            self.front.prev = new_node
            self.front = new_node
        self.count +=1 
    
    def append_rear(self, value):
        new_node = Node(value)
        if self.is_empty():
            self.front = self.rear = new_node
        else:
            new_node.prev = self.rear
            self.rear.next = new_node
            self.rear = new_node
        self.count += 1
    
    def pop_front(self):
        if self.is_empty():
            raise IndexError("Pop from front on empty deque")
        value = self.front.value
        self.front = self.front.next
        if self.front:
            self.front.prev = None
        else:
            self.rear = None  # deque is now empty
        self.count -= 1
        return value

    def pop_rear(self):
        if self.is_empty():
            raise IndexError("Pop from rear on empty deque")
        value = self.rear.value
        self.rear = self.rear.prev
        if self.rear:
            self.rear.next = None
        else:
            self.front = None  # deque is now empty
        self.count -= 1
        return value

    def peek_front(self):
        if self.is_empty():
            raise IndexError("Peek front on empty deque")
        return self.front.value

    def peek_rear(self):
        if self.is_empty():
            raise IndexError("Peek rear on empty deque")
        return self.rear.value

```

```python
dq = Deque()
dq.append_rear(10)
dq.append_front(20)
dq.append_rear(30)

print(dq.peek_front())  # Output: 20
print(dq.peek_rear())   # Output: 30

print(dq.pop_front())   # Output: 20
print(dq.pop_rear())    # Output: 30
print(dq.size())        # Output: 1
print(dq.is_empty())    # Output: False

print(dq.pop_front())   # Output: 10
print(dq.is_empty())    # Output: True
```

```python
class Deque:
    def __init__(self, initial_capacity=4):
        self.capacity = initial_capacity
        self.data = [None] * self.capacity
        self.front = 0
        self.rear = 0
        self.count = 0

    def is_empty(self):
        return self.count == 0

    def size(self):
        return self.count

    def _resize(self):
        new_capacity = self.capacity * 2
        new_data = [None] * new_capacity

        # Copy data in proper order
        for i in range(self.count):
            new_data[i] = self.data[(self.front + i) % self.capacity]

        self.data = new_data
        self.front = 0
        self.rear = self.count
        self.capacity = new_capacity

    def append_rear(self, value):
        if self.count == self.capacity:
            self._resize()
        self.data[self.rear] = value
        self.rear = (self.rear + 1) % self.capacity
        self.count += 1

    def append_front(self, value):
        if self.count == self.capacity:
            self._resize()
        self.front = (self.front - 1 + self.capacity) % self.capacity
        self.data[self.front] = value
        self.count += 1

    def pop_front(self):
        if self.is_empty():
            raise IndexError("Pop from front on empty deque")
        value = self.data[self.front]
        self.data[self.front] = None  # Optional: help garbage collection
        self.front = (self.front + 1) % self.capacity
        self.count -= 1
        return value

    def pop_rear(self):
        if self.is_empty():
            raise IndexError("Pop from rear on empty deque")
        self.rear = (self.rear - 1 + self.capacity) % self.capacity
        value = self.data[self.rear]
        self.data[self.rear] = None
        self.count -= 1
        return value

    def peek_front(self):
        if self.is_empty():
            raise IndexError("Peek front on empty deque")
        return self.data[self.front]

    def peek_rear(self):
        if self.is_empty():
            raise IndexError("Peek rear on empty deque")
        return self.data[(self.rear - 1 + self.capacity) % self.capacity]
```

```python
dq = Deque()

dq.append_rear(10)
dq.append_front(20)
dq.append_rear(30)

print(dq.peek_front())  # 20
print(dq.peek_rear())   # 30

print(dq.pop_front())   # 20
print(dq.pop_rear())    # 30
print(dq.pop_front())   # 10

print(dq.is_empty())    # True
```
