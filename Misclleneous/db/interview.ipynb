Write a Python program that takes an integer n as input and prints the numbers from 1 to n with the following rules: 
Print "tic" instead of the number if it is a multiple of 3. 
Print "tac" instead of the number if it is a multiple of 5. 
Print "tictac" if the number is a multiple of both 3 and 5. 
For all other numbers, print the number itself. 

```python
def printNum(n: int) -> None:
    for i in range(n):
        if i % 3 ==0 and i % 5 == 0:
            print("tictac")
        elif i % 3 == 0:
            print("tic")
        
        elif i % 5 == 0:
            print("tac")
        else:
            print(f"Num :: {i}")


printNum(15)
```

You are driving a little too fast, and a police officer stops you. Write a function to return one of 3 possible results: "No ticket", "Small ticket", or "Big Ticket". If your speed is 60 or less, the result is "No Ticket". If speed is between 61 and 80 inclusive, the result is "Small Ticket". If speed is 81 or more, the result is "Big Ticket". Unless it is your birthday (encoded as a boolean value in the parameters of the function) -- on your birthday, your speed can be 5 higher in all cases. 
 

```python
def drivingChallan(speed: int, birthday: bool):
    if birthday:
        speed += 5

    if speed <= 60:
        print("No Ticket")
    
    elif 61 <= speed <=80:
        print("Small")
    
    elif speed >=81:
        print("Big Ticket")

drivingChallan(71, True)


```
